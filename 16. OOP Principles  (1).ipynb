# Encapsulation in Python

**Definition:**  
Encapsulation is an Object-Oriented Programming concept that **hides the internal details of a class** and restricts direct access to some of its attributes. This helps **protect the data** and control how it is modified or accessed.

In Python, encapsulation is achieved by using **private variables** (prefix `__`) and **getter/setter methods**.

## Real-Life Example: Bank Account

A bank account contains sensitive information like the account balance.  
- You **cannot directly modify** the balance. (You cannot access private variable(__) directly outside the class)
- You can only interact with it via **controlled methods** like deposit or withdraw.

## Python Code Example

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance  # Private attribute

    # Getter method
    def get_balance(self):
        return self.__balance

    # Setter methods (controlled access)
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ${amount}. New balance: ${self.__balance}")
        else:
            print("Invalid deposit amount!")

    def withdraw(self, amount):
        if amount > 0 :
              self.__balance -= amount
            print(f"Withdrawn ${amount}. Remaining balance: ${self.__balance}")
        else:
            print("Invalid withdrawal amount or insufficient funds!")


# --- User Interaction ---
owner_name = input("Enter account owner's name: ")
initial_balance = float(input("Enter initial balance: "))

account = BankAccount(owner_name, initial_balance)

while True:
    print("\nChoose an action:")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Exit")
    choice = input("Enter choice (1-4): ")

    if choice == '1':
        amount = float(input("Enter amount to deposit: "))
        account.deposit(amount)
    elif choice == '2':
        amount = float(input("Enter amount to withdraw: "))
        account.withdraw(amount)
    elif choice == '3':
        print(f"Current balance: ${account.get_balance()}")
    elif choice == '4':
        print("Thank you! Goodbye.")
        break
    else:
        print("Invalid choice! Please try again.")

---

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance  # Private attribute

    # Getter method
    def get_balance(self):
        return self.__balance

    # Setter methods (controlled access)
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ${amount}. New balance: ${self.__balance}")
        else:
            print("Invalid deposit amount!")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrawn ${amount}. Remaining balance: ${self.__balance}")
        else:
            print("Invalid withdrawal amount or insufficient funds!")

# Usage
account = BankAccount("Alice", 1000)

account.deposit(500)
account.withdraw(300)
print(f"Current balance: ${account.get_balance()}")
print(account.__balance)

# Inheritance in Python

**Definition:**  
Inheritance is an Object-Oriented Programming (OOP) concept where a class (**child class / subclass**) can **inherit attributes and methods** from another class (**parent class / superclass**).  
- This allows **code reuse** and **hierarchical relationships** between classes.  
- The child class can also **override** or **extend** the parent class functionality.

## Real-Life Example: Animals

- **Parent class:** Animal (common features like eat and sleep)  
- **Child classes:** Dog, Cat (specific features like bark or meow)  

## Python Code Example

```python
# Parent Class
class Animal:
    def eat(self):
        print("This animal eats food.")

    def sleep(self):
        print("This animal sleeps.")

# Child Class 1
class Dog(Animal):
    def bark(self):
        print("Dog barks: Woof! Woof!")

# Child Class 2
class Cat(Animal):
    def meow(self):
        print("Cat meows: Meow! Meow!")

# Usage
dog = Dog()
cat = Cat()

dog.eat()    # Inherited from Animal
dog.bark()   # Dog's own method

cat.sleep()  # Inherited from Animal
cat.meow()   # Cat's own method

# Abstraction in Python

**Definition:**  
Abstraction is an **OOP concept** that hides the **internal implementation details** and **only shows the essential features** to the user.  
It helps in **reducing complexity** and **improving code maintainability**.

In Python, abstraction is achieved using **abstract classes** from the `abc` module.

## Key Points:

- Shows **what an object does**, not **how it does it**.  
- Uses **abstract classes** with **abstract methods**.  
- Cannot create an **instance of an abstract class** directly.  
- Forces subclasses to implement the abstract methods.

## Real-Life Example: Car

Think about a **car**:  

- You **press the accelerator** → car moves.  
- You **press the brake** → car stops.  
- You **turn the steering wheel** → car changes direction.  

You **don’t need to know** how the engine, brakes, or transmission work internally.  
The **car interface (accelerator, brake, steering wheel)** is **abstraction** — it shows **what the car can do**, not how it does it.


## Python Code Example
ABC stands for Abstract Base Class.
purpose to define abstract classes in Python.
Abstract class that cannot be instantiated directly and can have abstract methods.
Abstract methods must be implemented by any subclass.

```python
from abc import ABC, abstractmethod

# Abstract class
class Car(ABC):

    @abstractmethod
    def start_engine(self):
        pass

    @abstractmethod
    def stop_engine(self):
        pass

    @abstractmethod
    def drive(self):
        pass

# Concrete class
class Tesla(Car):

    def start_engine(self):
        print("Tesla engine started silently.")

    def stop_engine(self):
        print("Tesla engine stopped.")

    def drive(self):
        print("Tesla is driving automatically.")

# Another concrete class
class Toyota(Car):

    def start_engine(self):
        print("Toyota engine started with ignition sound.")

    def stop_engine(self):
        print("Toyota engine stopped.")

    def drive(self):
        print("Toyota is driving manually.")

# Using the classes
my_tesla = Tesla()
my_tesla.start_engine()   # Output: Tesla engine started silently.
my_tesla.drive()          # Output: Tesla is driving automatically.
my_tesla.stop_engine()    # Output: Tesla engine stopped.

my_toyota = Toyota()
my_toyota.start_engine()  # Output: Toyota engine started with ignition sound.
my_toyota.drive()         # Output: Toyota is driving manually.
my_toyota.stop_engine()   # Output: Toyota engine stopped.

# Polymorphism in Python

**Definition:**  
Polymorphism is an **Object-Oriented Programming (OOP) concept** that allows **objects of different classes** to be **used in the same way** through a **common interface**, while each object behaves differently.  

The word **Polymorphism** means **“many forms”**.  

## Key Points:

- One **method name**, **different behaviors**.  
- Helps in **code flexibility** and **reusability**.  
- Achieved in Python using:
  1. **Method Overriding** (in inheritance)  
  2. **Operator Overloading**  
  3. **Duck Typing**  


## Real-Life Example: Cars

Think of **different types of cars**:

- Every car has a **`start_engine()` method**, but the **way it starts differs**:  
  - Tesla: starts silently  
  - Toyota: starts with engine sound  
- Polymorphism allows us to **call `start_engine()` on any car object**, and each car responds differently.


## Python Code Example

```python
# Base class
class Car:
    def start_engine(self):
        print("Starting car engine...")

# Tesla inherits Car
class Tesla(Car):
    def start_engine(self):
        print("Tesla engine starts silently")

# Toyota inherits Car
class Toyota(Car):
    def start_engine(self):
        print("Toyota engine starts with sound")

# Function using polymorphism
def start_car_engine(car):
    car.start_engine()

# Using different car objects
my_tesla = Tesla()
my_toyota = Toyota()

start_car_engine(my_tesla)   # Output: Tesla engine starts silently
start_car_engine(my_toyota)  # Output: Toyota engine starts with sound


# Comparison of OOP Concepts in Python (Cars Example)

| Feature / Aspect       | Encapsulation | Inheritance | Abstraction | Polymorphism |
|------------------------|---------------|------------|-------------|--------------|
| **Definition**         | Hiding internal details and controlling access to data | Deriving a new class from an existing class to reuse code | Hiding complex details and showing only essential features | Same method/interface behaves differently for different objects |
| **Purpose**            | Protect data and restrict direct access | Code reusability and hierarchy | Reduce complexity and expose only necessary details | Flexibility: one interface, many forms |
| **Real-Life Car Example** | Car dashboard hides engine details; you just press buttons | Tesla inherits basic car properties like wheels, seats from Car | Remote: you press start/stop without seeing engine wiring | Press start_engine(): Tesla → silent, Toyota → engine sound |
| **Python Example**     | `self.__engine_status` (private attribute) | `class Tesla(Car): ...` | `class Car(ABC)` with `@abstractmethod` | `car.start_engine()` behaves differently for Tesla and Toyota |
| **Implementation**     | Private attributes (`__attribute`) and getters/setters | Subclassing (`class Child(Parent)`) | Abstract classes and methods (`ABC`, `@abstractmethod`) | Method overriding, duck typing, operator overloading |
| **User Interaction**   | User interacts with limited interface | User uses child class with inherited features | User sees only essential methods | User calls the same method for different objects without worrying about details |

# Method Overloading in Python

## Definition
**Method Overloading** means having **multiple methods with the same name** but **different parameters** in a class.  
In Python, we do **not have true method overloading** like Java or C++.  
Instead, it is achieved using:
- Default arguments
- Variable-length arguments (`*args`)

## Real-Life Example
A calculator can:
- Add 2 numbers
- Add 3 numbers
- Add many numbers  

The **method name is the same** (`add`), but the **inputs are different**.

## Method Overloading using Default Arguments

```python
class Calculator:
    def add(self, a=0, b=0, c=0):
        return a + b + c

calc = Calculator()

print(calc.add(10, 20))       # Two values
print(calc.add(10, 20, 30))   # Three values


# Method Overriding in Python

## Definition
**Method Overriding** occurs when a **child class** provides a **different implementation** of a method that already exists in the **parent class**.

The method must have:
- Same name  
- Same parameters  
- Same return type  

But the **behavior is different**.

## Real-Life Example
A **Vehicle** can have a method `start()`.  
- A **Car** starts with a key.  
- A **Bike** starts with a button.  

Both use `start()`, but their behavior is different.

## Python Code Example

```python
# Parent Class
class Vehicle:
    def start(self):
        print("Vehicle is starting")

# Child Class
class Car(Vehicle):
    def start(self):
        print("Car starts with a key")

# Another Child Class
class Bike(Vehicle):
    def start(self):
        print("Bike starts with a button")

# Usage
v = Vehicle()
c = Car()
b = Bike()

v.start()
c.start()
b.start()


## Difference Between Method Overloading and Method Overriding

| Feature | Method Overloading | Method Overriding |
|--------|------------------|------------------|
| Definition | Same method name with different parameters | Same method name with same parameters but different implementation |
| Occurs in | Same class | Parent and child classes |
| Inheritance required | ❌ No | ✅ Yes |
| Purpose | Increase readability | Change behavior of parent class |
| Parameters | Must be different | Must be same |
| Return type | Can be different | Must be same |
| Supported in Python | Using `*args` or default values | Using inheritance |
| Example | `add(a)`, `add(a,b)` | `start()` in Vehicle and Car |

# Multiple Inheritance in Python

## Definition
**Multiple Inheritance** is when a **child class inherits from more than one parent class**.

This allows a class to:
- Use methods and properties of **multiple classes**
- Combine behaviors into a single class

## Real-Life Example

Scenario: A smart device that is both a Camera and a Music Player
## Python Code Example

```python
# Parent class 1
class Camera:
    def take_photo(self):
        print("Photo captured!")

# Parent class 2
class MusicPlayer:
    def play_music(self):
        print("Playing music!")

# Child class inheriting both Camera and MusicPlayer
class SmartDevice(Camera, MusicPlayer):
    def device_info(self):
        print("I am a smart device with multiple features.")

# Using the SmartDevice
device = SmartDevice()
device.take_photo()     # From Camera
device.play_music()     # From MusicPlayer
device.device_info()    # From SmartDevice


## OOP Relationships in Python:

### 1. is-a Relationship
- **Definition:** Child class **is a type of parent class**
- **Implemented using:** Inheritance

```python
class Vehicle:
    def move(self):
        print("Vehicle can move")

class Car(Vehicle):  # Car "is a" Vehicle
    def honk(self):
        print("Car can honk")

c = Car()
c.move()  # inherited from Vehicle
c.honk()  # own method


## 2. has-a Relationship

- **Definition:** A class contains another class as a part of it.
- **Implemented using:** Composition / Aggregation

```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()  # Car "has a" Engine

    def drive(self):
        self.engine.start()
        print("Car is driving")

c = Car()
c.drive()


### Difference Between is-a and has-a in Python

| Feature                  | is-a Relationship                  | has-a Relationship                  |
|---------------------------|----------------------------------|-----------------------------------|
| **Definition**            | Child class **is a type** of parent class | Class **contains** another class as a part of it |
| **Implementation**        | Inheritance (single/multiple)    | Composition / Aggregation          |
| **Example**               | `Car` **is a** `Vehicle`         | `Car` **has an** `Engine`         |
| **Purpose**               | Reuse functionality of parent class | Use features of another class without inheriting |
